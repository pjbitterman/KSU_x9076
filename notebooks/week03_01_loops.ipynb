## Lists: create, index, slice

The **list** data structure is an ordered collection of values, written with square brackets and separated by commas. 

"Ordered" means every value has a position — a Python list remembers what order you put things in, and lets you fetch a value by its position, or **index**.

**Predict before you run:** Python indexes start at `0`, not `1`. Given the list below, guess what `county_names[0]` and `county_names[-1]` each print — is `-1` an error, or does it mean something?

```python
county_names = ["Portage", "Summit", "Stark", "Trumbull"]

print(county_names[0])
print(county_names[-1])
```

`county_names[0]` is `"Portage"`: the first item sits at index `0`, not `1`. 

`county_names[-1]` is `"Trumbull"`: negative indices count from the end, so `-1` always means "the last item" no matter how long the list is

## Slicing: a range of items at once

A **slice** pulls out more than one item at a time: `list[start:stop]` gives you everything from index `start` up to, but *not including*, index `stop`.

**Predict before you run:** `county_names` has four items, indices `0` through `3`. Guess what `county_names[1:3]` prints — how many items, and which ones?

```python
print(county_names[1:3])
```

Two items: `['Summit', 'Stark']` — index `1` and index `2`, stopping *before* index `3`. The "stop is exclusive" rule can be tricky

## Growing a list: `append` and `len`

Lists aren't fixed size. `append()` adds an item to the end; `len()` tells you how many items are currently in the list

**Predict before you run:** Guess what `len(county_names)` reports before the `append()` call runs, and what it reports after.

```python
print(len(county_names))

county_names.append("Geauga")
print(county_names)
print(len(county_names))
```

`len()` first reports `4`, then `5`. This is because `append()` added `"Geauga"` to the end. The command does NOT mutate the *existing* items; `append()` only  adds a new one after the last item.

## `for` loops: over a list

A `for` loop runs a block of code once per item in a collection, binding a variable name to each item in turn. No manual indexing required. 

**Predict before you run:** Guess what prints, and in what order, when the loop below runs.

```python
for name in county_names:
    print(name)
```

Five lines, one county per line, in  order — `for` loops walk a list front to back, one item at a time. `name` is just a variable; call it whatever's descriptive (`county`, `item`, `row` are all common), but it should say what *one item is*, not what the collection is called

## `for` loops: over a `range`

Sometimes you want to loop a specific number of times, or need the index itself, not just each list item. Accordingly, `range(n)` generates the whole numbers from `0` up to (not including) `n`. This is the same "stop is exclusive" rule as we saw with slicing.

**Predict before you run:** Guess how many lines `range(5)` prints, and what the first and last numbers are.

```python
for i in range(5):
    print(i)
```

What did you get? Where did it start? Stop?

A trick:

`range` also accepts a start and a step: `range(1, 11, 2)` counts `1, 3, 5, 7, 9` — start at 1, stop before 11, count by 2.

```python
for distance in range(1, 11, 2):
    print(distance)
```

## Conditionals inside a loop: `if` / `elif` / `else`

A loop that treats every item identically can be limited. If we add a conditional inside the loop body, and each item can be handled differently based on its own value.

**Predict before you run:** Guess which of the four distances get labeled `"short"`, `"medium"`, and `"long"`, given the two cutoffs below.

```python
for distance in buffer_distances:
    if distance < 200:
        category = "short"
    elif distance < 750:
        category = "medium"
    else:
        category = "long"
    print(f"{distance}m -> {category}")
```

Python checks conditions top to bottom and stops at the first one that's `True`. In this canse `500` never even gets compared to the `else`, because `distance < 750` already caught it. Order your `elif` conditions from most to least specific, or you'll get surprising results.

## Filtering a list

A common use of a conditional inside a loop: keep only the items that pass a test, and throw the rest away. The pattern is always the same. We first start with an empty list, loop the source list, `append()` when the condition is met.

**Predict before you run:** Guess which of `county_names` (`Portage, Summit, Stark, Trumbull, Geauga`) end up in `large_counties`, using "more than 6 characters" as the filter.

```python
# start with an empty list
large_counties = []

# then loop through the county names and add the large ones to the list
for name in county_names:
    if len(name) > 6:
        large_counties.append(name)

# print the list of large counties
print(large_counties)
```

What did you get?

The only two names longer than six characters are... ??? 

`large_counties` started empty and only grew when a name passed the test, whereas the original `county_names` list never changed. Building a *new* list this way, instead of trying to remove items from the original mid-loop, is the correct approach

## Loop + accumulate: sum and count

The same "start with an empty list, grow it inside the loop" idea works for more than filtering. Running totals and counts follow the same form: start at zero, add something on every relevant pass through the loop.

**Predict before you run:** Guess the total of `park_acres` before you run the cell. Then check your arithmetic.

```python
park_acres = [120, 45, 980, 15, 300]

total_acres = 0
for acres in park_acres:
    total_acres = total_acres + acres

print(f"total acres: {total_acres}")
```

What did you get?

`total_acres` starts at `0` and each pass through the loop adds one more park's acreage. By the last iteration it contains the running sum of everything seen so far. `total_acres = total_acres + acres` looks circular but isn't: the right side is fully calculated with the *old* value before the new one gets stored.

**Predict before you run:** Using the same `park_acres`, guess how many parks are over 100 acres before running the count below.

```python
large_park_count = 0
for acres in park_acres:
    if acres > 100:
        large_park_count += 1 # this is new syntax for large_park_count = large_park_count + 1

print(f"parks over 100 acres: {large_park_count}")
```

What do you get?

`large_park_count += 1` is shorthand for `large_park_count = large_park_count + 1`. You'll see `+=` a lot for once counting-inside-a-loop becomes second nature. Sum, count, and filter are really the same pattern with a different accumulator: a running number, a running number, or a growing list.

## A peek ahead: dictionaries

Lists aren't the only way to collect values, and a running sum or count isn't the only thing we commonly want to calculate. 

Python's other core collection is the **dictionary**. Wait, what's a dictionary?

A **dictionary** is a set of `key: value` pairs instead of an ordered sequence. 


**Predict before you run:** `park_size_counts` below starts with three keys, each pointing at `0`. Guess what `park_size_counts["medium"]` prints, and what you'd expect after the line `park_size_counts["medium"] += 1` runs.

```python
park_size_counts = {"small": 0, "medium": 0, "large": 0}

print(park_size_counts["medium"])
park_size_counts["medium"] += 1
print(park_size_counts["medium"])
print(park_size_counts)
```

What did you get?

Reading a dictionary value works by key instead of position: 
 
- `park_size_counts["medium"]` fetches whatever's stored at the key `"medium"`, similar to how `county_names[0]` fetches whatever's at index `0`
- `park_size_counts["medium"] += 1` updates that one key in place; the other two keys, but `"small"` and `"large"`, are untouched. 

The final `print(park_size_counts)` shows the whole dictionary: `{'small': 0, 'medium': 1, 'large': 0}`. Note the curly braces, not square brackets. Each entry is a `key: value` pair separated by a colon.

**Predict before you run:** The loop below walks `park_acres` and sorts each value into `"small"` (under 50 acres), `"medium"` (50 to 300), or `"large"` (over 300). This is the same idea as the `if`/`elif`/`else` from earlier, except this time it counts into a dictionary instead of just printing a label. Guess the final tally for each of the three keys.

```python
park_size_counts = {"small": 0, "medium": 0, "large": 0}

for acres in park_acres:
    if acres < 50:
        park_size_counts["small"] += 1
    elif acres <= 300:
        park_size_counts["medium"] += 1
    else:
        park_size_counts["large"] += 1

print(park_size_counts)
```

## Common bug #1: off-by-one

Off-by-one errors come from getting a loop's start or stop point wrong by exactly one position — usually because `range()`'s "stop is exclusive" rule got forgotten mid-loop.

**Predict before you run:** `county_names` currently has 5 items (indices `0`–`4`). The loop below starts at index `1` and tries to go through index `5`. Guess whether this runs cleanly or crashes... AND if it crashes, on which iteration.

```python
for i in range(1, len(county_names) + 1):
    print(county_names[i])
```

It crashes: `IndexError: list index out of range`, on the last iteration. `len(county_names)` is `5`, so `range(1, 6)` counts `1, 2, 3, 4, 5`. Index `5` doesn't exist on a 5-item list (its last valid index is `4`). 

The fix, if you need indices at all: `range(len(county_names))`, which correctly stops at `4`. Better still, when you don't specifically need the index number: skip indexing entirely and write `for name in county_names:` <-- that form can't be off-by-one, because there's no index to get wrong.

```python
for i in range(len(county_names)):
    print(county_names[i])
```

Fewer moving parts, nothing to miscount.

## Common bug #2: mutating a list while iterating

A second classic bug: removing items from a list *while* a `for` loop is walking it. The list shifts every time an item is removed, but the loop's position doesn't know that. It just keeps counting forward, and quietly skips whatever slid into the spot it already passed.

**Predict before you run:** The loop below removes every county name longer than 6 characters, one at a time, from the list it's iterating. 

Both `"Trumbull"` (8 characters) and `"Cuyahoga"` (8 characters) should qualify for removal. Guess whether both actually get removed.

```python
counties = ["Trumbull", "Cuyahoga", "Stark", "Summit"]

for name in counties:
    if len(name) > 6:
        counties.remove(name)

print(counties)
```

What, what? 

`"Cuyahoga"` is still there, even though it's 8 characters and should have been removed. 

Why? the loop checks index `0` (`"Trumbull"`), removes it, and the list shifts left.`"Cuyahoga"` slides into index `0`. But the loop's internal counter has already moved on to index `1`, which is now `"Stark"`. `"Cuyahoga"` gets skipped entirely, and never checked a second time. This bug is dangerous BECAUSE it doesn't crash — it just silently produces a wrong answer. The only way to notice is checking the result by hand.

The fix is the same idea as filtering, from above: never remove from the list you're looping over. Build a *new* list of the items you want to keep instead.

```python
counties = ["Trumbull", "Cuyahoga", "Stark", "Summit"]

short_counties = []
for name in counties:
    if len(name) <= 6:
        short_counties.append(name)

print(short_counties)
```

The answer?

`['Stark', 'Summit']` is correct this time, because `counties` never changes size mid-loop; only `short_counties` grows. 


**The lesson**

Whenever you catch yourself calling `.remove()` or `.pop()` on the list a loop is actively walking, stop and ask whether you should be building a fresh list instead.
