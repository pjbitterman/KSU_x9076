## `def`: naming a block of code

A function definition starts with `def`, a name, and a parenthesized list of **parameters**. Parameters areplaceholders for the values the function will need every time it's called.

Remember our whitespace conversations...

Everything indented underneath is the function's **body**, and it doesn't execute until the function is *called* by name

### Let's take a look

````python
def describe_park(name, acres):
    sq_miles = acres / 640
    print(f"{name}: {acres} acres ({sq_miles:.2f} sq mi)")

describe_park("Portage Lakes", 460)
describe_park("Nelson Ledges", 65)
describe_park("Mosquito Lake", 2183)
````

## Parameters with defaults, and return values

`describe_park` above always prints and never hands anything back. If you call it, the only thing you get is text on the screen. 

Two more pieces make functions considerably more useful: 
- a **default value** for a parameter, so callers can skip it when the common case is fine, 
- a **return value**, so the function's result can be stored, passed along, or reused

**Predict before you run:** `search_radius_report` below has a parameter with a default value, `radius_miles=1`. 

Guess what the first call prints, given that it doesn't supply a `radius_miles` argument at all. Then guess whether the second and third calls behave any differently from each other.

````python
def search_radius_report(place_name, radius_miles=1):
    """Report a search radius around a place.

    Args:
        place_name (str): the place being searched around.
        radius_miles (float): search radius, in miles. Defaults to 1.

    Returns:
        str: a one-line summary of the search radius.
    """
    return f"search radius {radius_miles} mi around {place_name}"

print(search_radius_report("Kent"))
print(search_radius_report("Kent", 5))
print(search_radius_report("Kent", radius_miles=0.5))
````

`search_radius_report("Kent")` used the default, `1`, since nothing else was supplied

`search_radius_report("Kent", 5)` supplied `5` **positionally**, or by the parameter's position in the parenthesis list

`search_radius_report("Kent", radius_miles=0.5)` supplied the same argument **by keyword**, naming the parameter directly; 

Also, `return` vs. `print`

**Predict before you run:** `park_totals` below takes a *list* of acreages instead of one park at a time, and returns two values at once (a count and a total) packed into a **tuple**. Guess what `park_count, total_acres = park_totals(...)` unpacks those two values into, and what the two print lines below it will read

````python
def park_totals(acreages):
    """Summarize a list of park acreages as a count and a total.

    Args:
        acreages (list of float): acreages for a group of parks.

    Returns:
        tuple: (count, total_acres) -- the number of parks and their combined acreage.
    """
    count = len(acreages)
    total_acres = sum(acreages)
    return count, total_acres

park_count, total_acres = park_totals([460, 65, 2183, 96.6, 450.5])
print(f"{park_count} parks, {total_acres} acres total")
print(park_totals([460, 65, 2183, 96.6, 450.5]))
````

## Docstrings

A function's name and parameters tell you *how* to call it. 

A **docstring** is a triple-quoted string as the very first line of the function body, and it tells you *what the function does*. 


Our convention: one summary line, a blank line, then an `Args:` block (one line per parameter) and, when the function returns something, a `Returns:` block. `search_radius_report` above already has one. 

Python treats it specially.
- Not printed when the function runs
- attached to the function object itself
- any tool that inspects your code, including `help()`, can find it

**Predict before you run:** what prints below? The whole docstring, or just the first line?

````python
print(search_radius_report.__doc__)
````

**What did we get?**

`help(search_radius_report)` in a real notebook shows the same text with a bit more framing (the function's signature above it). Get in the habit of writing this block for every function you keep past a "scratch" cell

````python
help(search_radius_report)
````

`help()` prints more than `.__doc__` alone: a `Help on function search_radius_report in module ...` banner, the function's full signature — `search_radius_report(place_name, radius_miles=1)`, defaults included — and then the same docstring body underneath, reformatted with consistent indentation. `.__doc__` hands back the raw triple-quoted string exactly as written; `help()` builds a small formatted report around it

## Scope: local vs. global

Variables you create inside a function only exist **inside that function**, for as long as it's running. 

This is a function's **local scope**.

A notebook has its own variables its **global scope**

**Predict before you run:** `buffer_distance` is set to `"1 Mile"` at the notebook level (global scope). 

The function below creates its *own* `buffer_distance`, set to `"500 Meters"`, and prints it. 

Guess what prints **after** the function call. Does the notebook's `buffer_distance` change, or not?

````python
buffer_distance = "1 Mile"

def show_buffer_distance():
    buffer_distance = "500 Meters"
    print(f"inside the function: {buffer_distance}")

show_buffer_distance()
print(f"outside the function, still: {buffer_distance}")
````

**What happened?**

Could this lead to unexpected behavior?

**Predict before you run:** The function below tries to do something that looks reasonable:

1. read the global `park_count`
2. add one to it
3. return the result — without passing `park_count` in as a parameter. 

Guess whether this works.

````python
park_count = 0

def add_park():
    park_count = park_count + 1
    return park_count

add_park()
````

What did we get this time?

Why?

Python decides, when it reads the whole function body ahead of time, that `park_count` is a local variable — because *somewhere* in the function, `park_count` gets assigned a value. 

So `park_count` is a local variable, and the function tries to *read* `park_count` before the local one has ever been given a value. 

The global `park_count = 0` from the notebook is irrelevant here; the function never sees it.

## Composing functions

Small functions become more useful once they start calling each other. `describe_park` doesn't have to do ALL its own work. Instead, it can rely on a second, smaller function that handles just one piece of the logic.

**Predict before you run:** `classify_park_size` below sorts a park into `"small"`, `"medium"`, or `"large"` by acreage, using the cutoffs `100` and `500`. 

Given the three parks from the first section (460, 65, and 2183 acres) guess each one's size class before you run the cell. 

This version of `describe_park` below rewrites is different than before: instead of doing the size math itself, it calls `classify_park_size` and uses whatever is returned


### Let's work on reading the flow together

````python
def classify_park_size(acres):
    """Classify a park's size as small, medium, or large.

    Args:
        acres (float): the park's area in acres.

    Returns:
        str: "small", "medium", or "large".
    """
    if acres < 100:
        return "small"
    elif acres < 500:
        return "medium"
    else:
        return "large"

def describe_park(name, acres):
    """Build a one-line summary of a park, including its size class.

    Args:
        name (str): the park's name.
        acres (float): the park's area in acres.

    Returns:
        str: a formatted summary combining the name, acreage, and size class.
    """
    size_class = classify_park_size(acres)
    return f"{name}: {acres} acres ({size_class})"

print(describe_park("Portage Lakes", 460))
print(describe_park("Nelson Ledges", 65))
print(describe_park("Mosquito Lake", 2183))
print(describe_park("Wingfoot Lake State Park", 96.6))
print(describe_park("Tinkers Creek", 450.5))
````

Let's break this down:
- What does the function do?

````python
def buffer_and_report(in_fc, out_fc, distance="1 Mile"):
    """Buffer in_fc by distance, save the result to out_fc, and summarize it.

    Args:
        in_fc (str): workspace-relative name or path of the input feature class.
        out_fc (str): workspace-relative name or path for the buffered output.
        distance (str): a linear distance arcpy accepts, e.g. "1 Mile" or "500 Meters".
            Defaults to "1 Mile".

    Returns:
        str: a one-line summary of what was buffered, and how many features came out.
    """
    arcpy.analysis.Buffer(in_fc, out_fc, distance)
    feature_count = int(arcpy.management.GetCount(out_fc)[0])
    return f"{in_fc} -> {out_fc}: {feature_count} features buffered at {distance}"
````

## Things don't always go well

Every arcpy tool call can fail for a lot of specific reasons, but almost all of them sort into three buckets:

**1. A missing file.** Typo a filename, or point at a workspace that doesn't have what you think it has, and the tool won't work:

```
ERROR 000732: Input Features: Dataset ohio_state_prks.shp does not exist or is not supported
Failed to execute (Buffer).
```

**2. A locked layer.** Point a tool at a feature class someone else (another Pro session, an open edit session) currently has locked, and the tool can't get the access it needs:

```
ERROR 000464: Cannot get exclusive schema lock. Either being edited or in use by another application.
Failed to execute (Buffer).
```

**3. A bad parameter.** Pass a value a parameter doesn't accept (an unrecognized unit, an out-of-range option_)_ and the tool rejects it before it ever touches your data:

```
ERROR 000824: The value is not a member of Linear Unit.
Failed to execute (Buffer).
```

## The fix: `try`/`except` around a tool call

**Try** the tool call, and if arcpy raises `arcpy.ExecuteError`, **catch** it instead of letting it stop the notebook. `arcpy.GetMessages(2)` (the `2` means "severity 2 only," arcpy's code for error-level messages) hands back the exact error text the tool produced, so you can print or log it instead of guessing

**Predict before you run:** A different issue than before, this time wrapped in `try`/`except`. Guess whether the notebook stops this time, and what prints if it doesn't

````python
try:
    arcpy.analysis.Buffer("ohio_state_prks.shp", "parks_1mi.shp", "1 Mile")
except arcpy.ExecuteError:
    print(arcpy.GetMessages(2))
````

**Expected result:**

```
ERROR 000732: Input Features: Dataset ohio_state_prks.shp does not exist or is not supported
Failed to execute (Buffer).
```

No traceback, no crash... the `except` block catches `arcpy.ExecuteError` before Python treats it as fatal, and `arcpy.GetMessages(2)` prints arcpy's own error text, the same message that would otherwise have shown up mid-traceback. `arcpy.GetMessages()` on its own returns *every* message a tool produced, including informational ones; passing `2` filters down to errors only, which is almost always what you want inside an `except` block

## Validating inputs first

`try`/`except` handles a failure *after* arcpy attempts the tool. 

Often it's better to check *before* you even try. You've seen `arcpy.Exists()` before, and pairing it with `raise ValueError(...)` lets you write your own, more specific message instead of waiting for arcpy's

````python
def validate_input(fc):
    """Raise ValueError if fc does not exist in the current workspace.

    Args:
        fc (str): workspace-relative name or path to check.

    Raises:
        ValueError: if arcpy.Exists(fc) is False.
    """
    if not arcpy.Exists(fc):
        raise ValueError(f"{fc} does not exist in {arcpy.env.workspace} -- check the name and the workspace.")

validate_input("ohio_state_prks.shp")
````

### Let's try something a bit different. With a buddy, break down the code...

````python
def validate_distance(distance, max_miles=10):
    """Raise ValueError if distance is not a positive number under a sane cap.

    Args:
        distance (float): a candidate buffer distance, in miles.
        max_miles (float): the largest distance a caller may request. Defaults to 10.

    Raises:
        ValueError: if distance is not greater than 0, or distance exceeds max_miles.
    """
    if distance <= 0:
        raise ValueError(f"{distance} is not a valid buffer distance -- must be greater than 0.")
    if distance > max_miles:
        raise ValueError(f"{distance} miles exceeds the {max_miles}-mile cap -- check for a units mistake.")

for candidate in [2, -1, 500]:
    try:
        validate_distance(candidate)
        print(f"{candidate}: OK")
    except ValueError as e:
        print(f"{candidate}: {e}")
````

## A loop that survives failure

Put everything from today together: 

- validate first
- wrap the tool call in `try`/`except`
- print progress as you go
- collect every failure instead of stopping at the first one

...this batch loop can survive a run where several files are broken, and still finish the rest, then tell you exactly what to go fix

### Let's break it down

````python
fcs_to_buffer = ["ohio_counties.shp", "ohio_state_parks.shp", "missing_layer.shp", "portage_munis.shp"]
failed = []

for i, fc in enumerate(fcs_to_buffer):
    print(f"{i+1}/{len(fcs_to_buffer)}: {fc}")
    if not arcpy.Exists(fc):
        print(f"  skipped -- {fc} does not exist")
        failed.append(fc)
        continue
    base_name = fc.replace(".shp", "")
    out_name = f"{base_name}_1mi.shp"
    try:
        arcpy.analysis.Buffer(fc, out_name, "1 Mile")
        print(f"  buffered -> {out_name}")
    except arcpy.ExecuteError:
        print(f"  failed -- {arcpy.GetMessages(2)}")
        failed.append(fc)
        continue

print()
if failed:
    print(f"{len(failed)} of {len(fcs_to_buffer)} failed: {failed}")
else:
    print("all feature classes processed successfully")
````
